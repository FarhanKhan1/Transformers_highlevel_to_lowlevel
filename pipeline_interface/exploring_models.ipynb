```python
from transformers import pipeline
```

```python
print("Loading model...")
classifier = pipeline("sentiment-analysis")
print("Ready!")
```

```python
while True:
    text = input(["Enter text (q to quit): "])

    if text.lower() == "q":
        break

    print(classifier(text))
```
