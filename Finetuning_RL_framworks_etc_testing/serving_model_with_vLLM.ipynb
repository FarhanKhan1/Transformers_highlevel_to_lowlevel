```python
import concurrent.futures
import time
import requests

prompt_template_test = """Below is an instruction that describes a task. Write a response that appropriately completes the request.

### Instruction:
{}

### Response:
"""

questions = [
    "What is the derivative of x^2 + 3x?",
    "Solve for x: 2x + 5 = 15",
    "What is the integral of 2x?",
    "Explain the Pythagorean theorem.",
    "What is the square root of 144?",
    "Simplify: (x+2)(x-2)",
    "What is the derivative of sin(x)?",
    "Find the area of a circle with radius 5.",
    "What is 15% of 200?",
    "Explain what a prime number is.",
]

def send_request(question):
    start = time.time()
    response = requests.post(
        "http://localhost:8000/v1/completions",
        json={
            "model": "Farhankhan1/deepseek-r1-distill-qwen-1.5b-reasoning-finetune",
            "prompt": prompt_template_test.format(question),
            "max_tokens": 512,
            "temperature": 0.6,
            "top_p": 0.95,
        },
    )
    elapsed = time.time() - start
    return question, elapsed, response.status_code

start_total = time.time()
with concurrent.futures.ThreadPoolExecutor(max_workers=10) as executor:
    results = list(executor.map(send_request, questions))
total_elapsed = time.time() - start_total

for question, elapsed, status in results:
    print(f"[{status}] {elapsed:.2f}s — {question}")

print(f"\nTotal wall-clock time for 10 concurrent requests: {total_elapsed:.2f}s")
```

```python
import concurrent.futures
import time
import requests

# Repeat questions to get ~150 requests
questions_150 = questions * 15  # 10 questions x 15 = 150

def send_request_light(question):
    start = time.time()
    try:
        response = requests.post(
            "http://localhost:8000/v1/completions",
            json={
                "model": "Farhankhan1/deepseek-r1-distill-qwen-1.5b-reasoning-finetune",
                "prompt": prompt_template_test.format(question),
                "max_tokens": 512,
                "temperature": 0.6,
                "top_p": 0.95,
            },
            timeout=120,
        )
        elapsed = time.time() - start
        return elapsed, response.status_code
    except Exception as e:
        return time.time() - start, str(e)

start_total = time.time()
with concurrent.futures.ThreadPoolExecutor(max_workers=150) as executor:
    results = list(executor.map(send_request_light, questions_150))
total_elapsed = time.time() - start_total

times = [r[0] for r in results]
statuses = [r[1] for r in results]

print(f"Requests: {len(questions_150)}")
print(f"Successes (200): {sum(1 for s in statuses if s == 200)}")
print(f"Failures: {sum(1 for s in statuses if s != 200)}")
print(f"Min time: {min(times):.2f}s")
print(f"Max time: {max(times):.2f}s")
print(f"Avg time: {sum(times)/len(times):.2f}s")
print(f"Total wall-clock time: {total_elapsed:.2f}s")
```
